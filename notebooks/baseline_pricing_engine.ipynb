```python
import pandas as pd
```

```python
df = pd.read_csv("../data/processed/sales_cleaned.csv")
df.head()
```

```python
df["original_revenue"] = df["Price"] * df["Units Sold"]
```

```python
def adjust_price(row):

    price = row["Price"]

    # Inventory rule
    if row["Inventory Level"] < 20:
        price = price * 1.15

    elif row["Inventory Level"] > 100:
        price = price * 0.90

    # Demand rule
    if row["Units Sold"] > 50:
        price = price * 1.10

    elif row["Units Sold"] < 20:
        price = price * 0.95

    return price
```

```python
df["adjusted_price"] = df.apply(adjust_price, axis=1)
```

```python
df[["Price","adjusted_price"]].head()
```

```python
df["new_revenue"] = df["adjusted_price"] * df["Units Sold"]
```

```python
original_revenue = df["original_revenue"].sum()
new_revenue = df["new_revenue"].sum()

print("Original Revenue:", original_revenue)
print("New Revenue:", new_revenue)
```

```python
revenue_lift = abs((new_revenue - original_revenue) / original_revenue) * 100

print("Revenue Lift (%):", round(revenue_lift, 2))
```
